### رشته‌های پایتون
>دنباله‌ای از کاراکترهاست (string) در پایتون، رشته
>
> ما در پایتون برای نمایش رشته از تک کوتیشن یا دابل کوتیشن استفاده می‌کنیم. برای مثال

```python
# create a string using double quotes
string1 = "Python programming"

# create a string using single quotes
string1 = 'Python programming'
```

```python
# create string type variables

name = "Python"
print(name)

message = "I love Python."
print(message)
```

### دسترسی به کاراکترهای رشته در پایتون

![python_slices](../images/python-slices.png)

> Syntax: `[start:stop:step]`

### ما می‌توانیم به سه روش به کاراکترهای یک رشته دسترسی پیدا کنیم

> اندیس‌گذاری: یک روش این است که با رشته‌ها مانند یک لیست رفتار کنیم و از مقادیر اندیس استفاده کنیم. برای مثال

```python
greet = 'hello'

# access 1st index element
print(greet[1]) # "e"
```

> اندیس‌گذاری منفی: پایتون، مشابه لیست‌ها، امکان اندیس‌گذاری منفی را برای رشته‌ها نیز فراهم می‌کند. برای مثال

```python
greet = 'hello'

# access 4th last element
print(greet[-4]) # "e"
```

> با استفاده از عملگر برش‌دهی (یعنی علامت دونقطه `:`) به دامنه‌ای از کاراکترها در یک رشته دسترسی پیدا کنید. برای مثال :(Slicing) برش‌دهی

```python
greet = 'Hello'

# access character from 1st index to 3rd index
print(greet[1:4])  # "ell"
```

```python
string = "Python programming"
print(string[0])
print(string[-1])
print(string[1:4])
print(string[:2])
print(string[-4:-1])
print(string[1:5:2])
```

### رشته‌های پایتون تغییرناپذیر هستند
> در پایتون، رشته‌ها تغییرناپذیر هستند. این بدان معناست که کاراکترهای یک رشته قابل تغییر نیستند. برای مثال

```python
message = 'Hola Amigos'
message[0] = 'H'
print(message)
```

> با این حال، می‌توانیم نام متغیر را به یک رشته جدید اختصاص دهیم. برای مثال

```python
message = 'Hola Amigos'

# assign new string to message variable
message = 'Hello Friends'

print(message); # prints "Hello Friends"
```

### رشته چندخطی در پایتون
> در پایتون می‌توانیم رشته‌های چندخطی نیز ایجاد کنیم. برای این کار، از سه علامت نقل‌قول دوتایی (یعنی """) یا سه علامت نقل‌قول تکی (یعنی ''') استفاده می‌کنیم. برای مثال

```python
# multiline string 
message = """
Never gonna give you up
Never gonna let you down
"""

print(message)
```

### اتصال دو یا چند رشته
> در پایتون، می‌توانیم با استفاده از عملگر `+` دو یا چند رشته را به هم متصل (الحاق) کنیم

```python
greet = "Hello, "
name = "Jack"

# using + operator
result = greet + name
print(result)

# Output: Hello, Jack
```

### طول رشته در پایتون
>برای یافتن طول یک رشته استفاده می‌کنیم. برای مثال `len()` در پایتون، از متد

```python
greet = 'Hello'

# count length of greet string
print(len(greet))

# Output: 5
```

> **ابزار پاک‌سازی ثبت‌نام دانشجویان**
>
> یک مؤسسه آموزشی اطلاعات دانشجویان را با قالب‌بندی نامنظم دریافت می‌کند. وظیفه شما پاک‌سازی داده‌ها و تهیه یک گزارش ثبت‌نام با قالب‌بندی مناسب است
>
* raw_name = "   brian griffen   "
  
* raw_course = "python for machine learning"

* raw_email = "BRIAN.GRIFFEN@EXAMPLE.COM "

* raw_city = "  london "

* skills = "python,sql,git,machine learning"

----------------------------------------------------------------------

**خروجی مورد انتظار**

---------- STUDENT REPORT ----------

* Name           : Brian Griffen

* Course          : PYTHON FOR DEEP LEARNING

* Email           : brian.griffen@example.com

* City            : London

* Name length     : 13

* Letter 'a' count: 1

* Starts with 'Brian': True

* Ends with '.com'    : True

* First 3 letters of name : Bri

* Skills           : python | sql | git | machine learning

> Bonus: find username
>
> * Username: brian.griffen

```python
raw_name = "   brian griffen   "
raw_course = "python for machine learning"
raw_email = "BRIAN.GRIFFEN@EXAMPLE.COM "
raw_city = "  london "
skills = "python,sql,git,machine learning"

# 1. Clean the text
name = raw_name.strip()
email = raw_email.strip()
city = raw_city.strip()

# 2. Format the text
name = name.title()
course = raw_course.upper()
email = email.lower()
city = city.title()

# 3. Replace text
course = course.replace("MACHINE", "DEEP")

# 4. Count characters
a_count = name.lower().count("a")

# 5. Find a character
at_position = email.find("@")

# 6. Check text
ends_with_com = email.endswith(".com")
starts_with_brian = name.startswith("Brian")

# 7. Split and join
skills_list = skills.split(",")
skills_joined = " | ".join(skills_list)

# 8. Length
name_length = len(name)

# 9. String slicing
first_three = name[:3]

# 10. Final report with f-string
print("---------- STUDENT REPORT ----------")
print(f"Name            : {name}")
print(f"Course          : {course}")
print(f"Email           : {email}")
print(f"City            : {city}")
print(f"Name length     : {name_length}")
print(f"Letter 'a' count: {a_count}")
print(f"Starts with 'Brian': {starts_with_brian}")
print(f"Ends with '.com'    : {ends_with_com}")
print(f"First 3 letters : {first_three}")
print(f"Skills          : {skills_joined}")
print("------------------------------------")

# Bonus challenge (find username using slicing)
username = email[:at_position]
print(f"Username: {username}")
```
