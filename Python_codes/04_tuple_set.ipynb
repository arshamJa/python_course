### تاپل پایتون
> (Tuple) تاپل
>
>مجموعه‌ای شبیه به لیست در پایتون است. تفاوت اصلی این است که پس از ایجاد یک تاپل، نمی‌توان آن را تغییر داد




### ایجاد یک تاپل در پایتون
> ما با قرار دادن عناصر درون پرانتز () یک تاپل ایجاد می‌کنیم. برای مثال

```python
numbers = (1, 2, -5)
print(numbers)
```

>یک تاپل ایجاد کنیم. برای مثال `tuple()` ما همچنین می‌توانیم با استفاده از سازنده‌ی

```python
tuple_constructor = tuple(('Jack', 'Maria', 'David'))
print(tuple_constructor)
```

>که می‌توانیم در پایتون ایجاد کنیم، آمده است (tuples) در اینجا انواع مختلف تاپل‌هایی

```python
# create an empty tuple
empty_tuple = ()
print(empty_tuple)

# tuple of string types
names = ('James', 'Jack', 'Eva')
print (names)

# tuple of float types
float_values = (1.2, 3.4, 2.1)
print(float_values)

# tuple including string and integer
mixed_tuple = (2, 'Hello', 'Python')
print(mixed_tuple)
```

### ویژگی‌های تاپل
* مرتب‌شده - آن‌ها ترتیب عناصر را حفظ می‌کنند
* تغییرناپذیر - پس از ایجاد، قابل تغییر نیستند
* اجازه دادن به مقادیر تکراری - می‌توانند شامل مقادیر تکراری باشند

### دسترسی به عناصر تاپل
> نامیده می‌شود (index) با عددی مرتبط است که اندیس(tuple) هر عضو در یک تاپل 
>
> اندیس همیشه از ۰ شروع می‌شود؛ بدین معنا که اولین عضو تاپل در اندیس ۰، دومین عضو در اندیس ۱ و به همین ترتیب قرار دارد


```python
languages = ('Python', 'Swift', 'C++')

# access the first item
print(languages[0])  

# access the third item
print(languages[2])   
```

### تاپل قابل تغییر نیست
> هستند (immutable)  تاپل‌های پایتون تغییرناپذیر
>
> ما نمی‌توانیم آیتم‌های یک تاپل را اضافه، تغییر یا حذف کنیم
>
> اگر سعی کنیم یک تاپل را تغییر دهیم، با خطا مواجه خواهیم شد. برای مثال 

```python
cars = ('BMW', 'Tesla', 'Ford', 'Toyota')

# trying to modify a tuple
cars[0] = 'Nissan'    # error
print(cars)
```

### طول تاپل در پایتون
> `len()` تابع
>
> برای یافتن تعداد عناصر موجود در یک تاپل استفاده می‌کنیم. برای مثال

```python
cars = ('BMW', 'Tesla', 'Ford', 'Toyota')
print('Total Items:', len(cars)) 
```

### متدهای تاپل در پایتون
* Python Tuple `count()`
> تعداد دفعات تکرار عنصر مشخص‌شده در تاپل را برمی‌گرداند `count()`متد
> 
* Python Tuple `index()`
> اندیس عنصر مشخص‌شده در تاپل را برمی‌گرداند `index()` متد 

```python
# tuple of vowels
vowels = ('a', 'e', 'i', 'o', 'i', 'u')

# counts the number of i's in the tuple
count = vowels.count('i')

print(count)
```

```python
# tuple containing vowels
vowels = ('a', 'e', 'i', 'o', 'u')

# index of 'e' in vowels
index = vowels.index('e')

print(index)
```

> : به‌طور کلی، زمانی باید از تاپل‌ها استفاده کنید که نیاز دارید
*  تضمین یکپارچگی داده‌ها: تاپل‌ها  تغییرناپذیر هستند؛ بدین معنا که پس از ایجاد، نمی‌توانید عناصر آن‌ها را تغییر دهید. این ویژگی تغییرناپذیری، ثبات داده‌ها را تضمین کرده و باعث می‌شود مقادیر موجود در تاپل بدون تغییر باقی بمانند
* کاهش مصرف حافظه: تاپل‌ها در مقایسه با لیست‌ها سربار حافظه کمتری دارند، زیرا مقدار ثابتی از حافظه را به خود اختصاص می‌دهند. این ویژگی به‌ویژه هنگام کار با مجموعه‌های بزرگ داده یا در محیط‌هایی با محدودیت حافظه، مزیت محسوب می‌شود
* بهبود عملکرد: تاپل‌ها معمولاً از نظر ایجاد، پیمایش و دسترسی به عناصر، کارآمدتر از لیست‌ها هستند. این ویژگی می‌تواند منجر به بهبود عملکرد شود، به‌ویژه هنگام کار با مجموعه‌داده‌های حجیم

### مجموعه‌ها در پایتون
> (Set) مجموعه
>
> به معنای گردایه‌ای از داده‌های منحصربه‌فرد است؛ بدین معنا که عناصر درون یک مجموعه نمی‌توانند تکراری باشند
>
> برای مثال، اگر نیاز به ذخیره اطلاعات مربوط به شماره‌های دانشجویی داشته باشیم، استفاده از مجموعه مناسب است؛ زیرا شماره‌های دانشجویی نمی‌توانند تکراری باشند


### در پایتون (Set) ایجاد یک مجموعه 
> در پایتون، ما مجموعه‌ها را با قرار دادن تمام عناصر درون آکولاد `{}` و جدا کردن آن‌ها با ویرگول ایجاد می‌کنیم
>
> می‌تواند شامل هر تعداد عضو باشد و این اعضا می‌توانند از انواع مختلفی (مانند عدد صحیح، عدد اعشاری، تاپل، رشته و غیره) باشند (Set) یک مجموعه
>
> نظیر لیست‌ها، مجموعه‌ها یا دیکشنری‌ها باشد (mutable) اما یک مجموعه نمی‌تواند شامل عناصر تغییرپذیر 

```python
# create a set of integer type
student_id = {112, 114, 116, 118, 115}
print('Student ID:', student_id)

# create a set of string type
vowel_letters = {'a', 'e', 'i', 'o', 'u'}
print('Vowel Letters:', vowel_letters)

# create a set of mixed data types
mixed_set = {'Hello', 101, -2, 'Bye'}
print('Set of mixed data types:', mixed_set)
```

> در مثال بالا، با قرار دادن تمام عناصر درون آکولاد {}، انواع مختلفی از مجموعه‌ها را ایجاد کرده‌ایم 
>
> نکته: هنگام اجرای این کد، ممکن است خروجی را با ترتیبی متفاوت دریافت کنید؛ زیرا مجموعه ترتیب خاصی ندارد

### ایجاد یک مجموعه خالی در پایتون
> ایجاد یک مجموعه خالی کمی چالش‌برانگیز است؛ چرا که استفاده از آکولادهای خالی `{}` در پایتون، یک دیکشنری خالی ایجاد می‌کند
>
> بدون هیچ آرگومانی استفاده می‌کنیم. برای مثال `set()` برای ایجاد یک مجموعه بدون هیچ عضوی، از تابع

```python
# create an empty set
empty_set = set()

# create an empty dictionary
empty_dictionary = { }

# check data type of empty_set
print('Data type of empty_set:', type(empty_set))

# check data type of dictionary_set
print('Data type of empty_dictionary:', type(empty_dictionary))
```

### در پایتون (Set) افزودن و به‌روزرسانی عناصر مجموعه
> مجموعه‌ها تغییرپذیر هستند؛ با این حال، از آنجا که فاقد ترتیب هستند، اندیس‌گذاری در آن‌ها معنایی ندارد
>
> ما نمی‌توانیم با استفاده از نمایه‌سازی یا برش‌دهی به یکی از عناصر مجموعه دسترسی پیدا کنیم یا آن را تغییر دهیم؛ چرا که نوع داده‌ایِ مجموعه  از این قابلیت پشتیبانی نمی‌کند

### افزودن عناصر به یک مجموعه در پایتون
>  `add()` از متد
>
> استفاده می‌کنیم. برای مثال (set) در پایتون، برای افزودن یک عضو به یک مجموعه

```python
numbers = {21, 34, 54, 12}

print('Initial Set:',numbers)

# using add() method
numbers.add(32)

print('Updated Set:', numbers) 
```

### در پایتون (Set) به‌روزرسانی مجموعه
> `update()` متد
>
> با عناصر موجود در سایر انواع مجموعه‌ها (مانند لیست‌ها، تاپل‌ها، مجموعه‌ها و غیره) استفاده می‌شود. برای مثال، (set) برای به‌روزرسانی یک مجموعه

```python
companies = {'Lacoste', 'Ralph Lauren'}
tech_companies = ['apple', 'google', 'apple']

# using update() method
companies.update(tech_companies)

print(companies)
```

### حذف یک عنصر از مجموعه
> `discard()` ما از متد
>
> برای حذف عنصر مشخص شده از یک مجموعه استفاده می‌کنیم. برای مثال

```python
languages = {'Swift', 'Java', 'Python'}

print('Initial Set:',languages)

# remove 'Java' from a set
removedValue = languages.discard('Java')

print('Set after remove():', languages)
```

### عملیات مجموعه در پایتون
> Set نوع داده‌ای
>
> در پایتون، متدهای داخلی متنوعی را برای انجام عملیات ریاضی مجموعه‌ها — مانند اجتماع، اشتراک، تفاضل و تفاضل متقارن — فراهم می‌کند

### اجتماع دو مجموعه
> است B و A شامل تمام اعضای مجموعه‌های B و A اجتماع دو مجموعه

![set_union](../images/set_union.png)

>برای انجام عملیات اتحاد مجموعه‌ها استفاده می‌کنیم. برای مثال `union()` ما از عملگر `|` یا متد 

```python
# first set
A = {1, 3, 5}

# second set
B = {0, 2, 4}

# perform union operation using |
print('Union using |:', A | B)

# perform union operation using union()
print('Union using union():', A.union(B)) 
```

### اشتراک مجموعه‌ها
> است A و B شامل عناصر مشترک بین مجموعه A و B اشتراک دو مجموعه

![set_intersection](../images/set_intersection.png)

>`intersection()` از عملگر `&` یا متد
>
> در پایتون برای انجام عملیات اشتراک مجموعه‌ها استفاده می‌کنیم برای مثال

```python
# first set
A = {1, 3, 5}

# second set
B = {1, 2, 3}

# perform intersection operation using &
print('Intersection using &:', A & B)

# perform intersection operation using intersection()
print('Intersection using intersection():', A.intersection(B)) 
```

### تفاوت بین دو مجموعه
> وجود ندارند B است که در مجموعه A شامل عناصری از مجموعهA A و B تفاضل دو مجموعه

![set_diff](../images/set_diff.png)

>`difference()` از عملگر`-` یا متد
>
> برای محاسبه تفاضل بین دو مجموعه استفاده میکنیم

```python
# first set
A = {2, 3, 5}

# second set
B = {1, 2, 6}

# perform difference operation using &
print('Difference using &:', A - B)

# perform difference operation using difference()
print('Difference using difference():', A.difference(B)) 
```

### تفاضل متقارن مجموعه‌ها
> به جز اعضای مشترک آن‌هاست A و B شامل تمام اعضای A و B تفاضل متقارن بین دو مجموعه

![set_symmetric_diff](../images/set_symmetric_diff.png)

>`^` یا `symmetric_difference()` در پایتون، ما از عملگر
>
>برای انجام تفاضل‌های متقارن بین دو مجموعه استفاده می‌کنیم. برای مثال  یا متد

```python
# first set
A = {2, 3, 5}

# second set
B = {1, 2, 6}

# perform difference operation using &
print('using ^:', A ^ B)

# using symmetric_difference()
print('using symmetric_difference():', A.symmetric_difference(B)) 
```

### *سیستم حضور و غیاب*

**داستان:** 
آقای بنر یک کلاس برنامه‌نویسی پایتون با ۱۰ دانش‌آموز برگزار می‌کند. او هر روز صبح، ۱۵ دقیقه را صرف بررسی حضور و غیاب با استفاده از یک دفتر ثبت کاغذی می‌کند. گاهی اوقات او به‌اشتباه نام دانش‌آموز دیگری را علامت می‌زند، نام‌ها را تکرار می نویسد یا فراموش می‌کند که غایبان را بشمارد

```python
# Permanent class list (Tuple)
class_list = ( "Alice", "Bob", "Charlie", "David", "Emma", "Frank", "Grace", "Henry")

# Students who attended today (Set)
present = {"Alice", "Charlie", "Emma", "Grace", "Henry"}

# Convert tuple to set for calculations
all_students = set(class_list)

# Find absent students
absent = all_students - present

# Display results
print("===== Attendance Report =====")

print("\nClass List:")
print(class_list)

print("\nPresent Students:")
print(present)

print("\nAbsent Students:")
print(absent)

print("\nTotal Students:", len(class_list))
print("Present:", len(present))
print("Absent:", len(absent))

attendance_percentage = (len(present) / len(class_list)) * 100

print("\nAttendance Percentage:", attendance_percentage, "%")
```
