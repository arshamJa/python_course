### Lambda function
> **تابع لامبدا چیست؟**
>
> تابع لامبدا یک تابع کوچک و تک‌خطی است که بدون نام تعریف می‌شود
>
> این توابع می‌توانند هر تعداد پارامتر ورودی داشته باشند، اما تنها شامل یک عبارت هستند؛ نتیجه‌ی همین عبارت نیز به عنوان خروجی تابع بازگردانده می‌شود

![lambda](../images/lambda.png)

[Reference: Medium](https://jegananbalagan03.medium.com/lambda-function-in-python-05a0626cc67d)

```python
def short_function(x):
    return x * 2

# Rather than using function we can use Lambda
equiv_anon = lambda x: x * 2
```

```python
x = lambda a, b : a * b
print(x(5, 6))
```

```python
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(check(5))   
print(check(-3))  
print(check(0))
```

### **map()**
> این تابع، یک تابع مشخص را برای تک‌تک عناصر یک شیء قابل‌تکرار (مانند لیست‌ها، تاپل‌ها و غیره) اجرا می‌کند
>
> Syntax: map(function, iterables)
>
>: این تابع دو آرگومان می‌گیرد
>
> * `function`
>
> تابعی که بر تک‌تک عناصر یک شیء قابل‌تکرار اعمال می‌شود
>
> * `iterables`
>
> اشیاء قابل‌تکرار مانند لیست‌ها، تاپل‌ها و غیره

### Loop نسبت به Map مزایای استفاده از
* map()
* هنگام کار با مجموعه‌های بزرگ، معمولاً سریع‌تر عمل می‌کند
* loops
* معمولاً کندتر هستند، زیرا مستلزم آن‌اند که مفسر دستور حلقه را برای هر تکرار اجرا کند

```python
numbers = [1,2,3,4]

# returns the square of a number
def square(number):
  return number * number

# apply square() to each item of the numbers list
squared_numbers = map(square, numbers)

# converting to list for printing
result = list(squared_numbers)
print(result)
```

### **map() with Lambda**

```python
numbers = (1, 2, 3, 4)
result = map(lambda x: x*x, numbers)
print(result)

# convert to list and print it
print(list(result))
```

```python
ages = [25, 32, 45, 60, 75]
age_bins = list(map(lambda x: 'young' if x < 35 else ('middle-aged' if x < 60 else 'senior'), ages))
print(age_bins)
```

> **filter()**
>
> در پایتون، از این تابع برای جدا کردن (فیلتر کردن) عناصر یک شیء قابل‌تکرار (مانند لیست، تاپل و غیره) بر اساس یک شرط استفاده می‌شود
>
>  این تابع، یک تابع دیگر را روی تک‌تک عناصر آن مجموعه اعمال می‌کند و تنها عناصری در نتیجه نهایی قرار می‌گیرند که تابع مذکور برای آن‌ها مقدار درست را برگرداند.
>
> Syntax: filter(function, iterables)
>
>: این تابع دو آرگومان می‌گیرد
>
> * `function`
> *  تابعی که برای هر عنصر در شیء قابل‌تکرار مقدار درست یا غلط را برمی‌گرداند
> * `iterable`
> * آن شیء قابل‌تکرار (مانند لیست یا تاپل) است که می‌تواند عناصرش را فیلتر کند

```python
numbers = [1, 2, 3, 4, 5, 6, 7]

def is_even(x):
    return x % 2 == 0
    
even_numbers = filter(is_even, numbers)
print(list(even_numbers))
```

### **filter() with a Lambda**

```python
numbers = [1, 2, 3, 4, 5, 6, 7]

even_numbers = list(filter(lambda x: x % 2 == 0 , numbers))
print(even_numbers)
```
