> **ASCII** (استاندارد آمریکایی برای تبادل اطلاعات)
>
> یک استاندارد کدگذاری نویسه ۷-بیتی است که ۱۲۸ نویسه را نمایش می‌دهد
>
> شامل حروف انگلیسی، ارقام، علائم نگارشی و نویسه‌های کنترلی

Reference: [Real python website](https://realpython.com/ref/glossary/ascii/#:~:text=character%20encoding%20standard.-,Character%20Set,DEL%20(127)%3A%20Delete%20character)

> **تبدیل کاراکتر به عدد**
>
> `ord()` تابع
>
> یک رشته شامل یک کاراکتر را دریافت کرده و مقدار عددی(صحیح) آن را برمی‌گرداند

```python
char_value = 'g'
ascii_code = ord(char_value)
print(ascii_code)
```

> **تبدیل عدد به کاراکتر**
>
 >`char()` تابع
 >
 > یک عدد صحیح را دریافت کرده و رشتهٔ کاراکتریِ متناظر با آن را برمی‌گرداند

```python
ascii_code = 103
char_value = chr(ascii_code)
print(char_value)
```

> :ASCII تبدیل کل یک رشته به
>
> برای دریافت کدهای مقادیر تمام کاراکترهای یک رشته می‌توانید از
>
>  استفاده کنید (list comprehension)

```python
python_text = "Hello"
ascii_list = [ord(char) for char in python_text]
print(ascii_list)
```
