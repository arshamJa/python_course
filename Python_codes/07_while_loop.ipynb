> را به تکرار بعدی منتقل کنید و از باقی‌مانده‌ی بدنه حلقه صرف‌نظر نمایید `for` می‌توانید اجرای حلقه **continue** با استفاده از کلمه کلیدی

```python
for a in [1, 2, 3, 4, 5, 6]:
    if a == 4:
        continue        # this will return to the loop without going for print
    print(a)
```

> می‌گذرد `None` به این قطعه کد توجه کنید که اعداد صحیح موجود در یک لیست را با هم جمع می‌کند و از مقادیر

```python
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value

print(total)
```

> خارج شد for می‌توان به‌طور کامل از یک حلقه **break** با استفاده از کلمه کلیدی 

```python
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
total_until_5 = 0
for value in sequence:
    if value == 5:
        break
    total_until_5 += value

print(total_until_5)
```

### `while()` حلقه‌ 
> باشد، استفاده می‌شود (True) برای اجرای یک بلوک کد تا زمانی که شرط مشخص‌شده برقرار while در پایتون، از حلقه

```python
number = 1
while number <= 3:
    print(number)
    number = number + 1  # number += 1
```

```python
while True:
    user_input = input('Enter your name: ')
    # terminate the loop when user enters end
    if user_input == 'end':
        print(f'The loop is ended')
        break
    print(f'Hi {user_input}')
```

### در پایتون while و حلقه for تفاوت بین حلقه 
* **For loop:** زمانی استفاده می‌شود که تعداد دفعات تکرار را می‌دانیم؛ اغلب همراه با لیست‌ها، تاپل‌ها، مجموعه‌ها یا دیکشنری‌ها به کار می‌رود

* **While loop:** زمانی استفاده می‌شود که تنها یک شرط پایان داریم و دقیقاً نمی‌دانیم [عملیات] چند بار تکرار خواهد شد

>شدن شرط حلقه اجرا می‌شود. برای مثال (False) اختیاری داشته باشد که پس از نادرست `else` می‌تواند یک عبارت  `while` در پایتون، حلقه

```python
counter = 0
while counter  <  2:
    print('This is inside loop')
    counter = counter + 1
else:
    print('This is inside else block')
```

### Rock/Paper/Scissors mini project

```python
import random

user_wins = 0
computer_wins = 0

options = ['rock', 'paper', 'scissors']

while True:
    user_input = input('Type Rock/Paper/Scissors or Q to quit: ').lower()
    if user_input == 'q':
        break
    if user_input not in options:
        continue

    random_number = random.randint(0, 2)
    # rock = 0, paper = 1, scissors = 2
    computer_choice = options[random_number]
    print(f'computer choice: {computer_choice}')

    if user_input == 'rock' and computer_choice == 'scissors':
        print('You win!')
        user_wins += 1
    elif user_input == 'paper' and computer_choice == 'rock':
        print('You win!')
        user_wins += 1
    elif user_input == 'scissors' and computer_choice == 'paper':
        print('You win!')
        user_wins += 1
    else:
        print('You lost')
        computer_wins += 1


print(f'You won {user_wins} times. computer won {computer_wins} times.')
print('Goodbye')
```
