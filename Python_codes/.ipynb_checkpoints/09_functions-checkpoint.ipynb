### return دستور 
> مقداری را از تابع برمی‌گردانیم return ما با استفاده از دستور 

```python
# function definition
def find_square(num):
    result = num * num
    return result

# function call
square = find_square(3)

print('Square:', square)
```

> ایجاد کرده‌ایم `find_square()` در مثال بالا، تابعی به نام
>
> این تابع یک عدد را دریافت کرده و مجذور آن را برمی‌گرداند
>
>
>اجرا نمی‌شود return همچنین نشان‌دهنده پایان یافتن تابع است. هیچ کدی پس از return نکته: دستور 

```python
# Exercise:
# count characters
def count_char(n, c):
    """
    :param n: is the name
    :param c: is the character
    :return: the number of character
    """
    count = 0
    for char in n:
        if char == c:
            count += 1
    return count

name = 'peter griffen'
result = count_char(name, 'e')
print(result)
```

### Keyword Argument (named arguments)

> Keyword arguments mean you pass values by parameter names while calling the function.
>
> * The order does not matter as long as the names are correct.
>
> * Defaults are used if some arguments are not provided.
>
> * This makes the code more readable and less error-prone.

```python
def nameAge(name, age):
    print("Hi, I am", name)
    print("My age is ", age)

nameAge(name="Prince", age=20)
print('--'*10)
nameAge(age=20, name="Prince")
```

### Positional Arguments (آرگومان‌های موقعیتی)
> آرگومان‌های موقعیتی بدین معناست که مقادیر دقیقاً به همان ترتیبی که پارامترها در تابع تعریف شده‌اند، به آن ارسال می‌شوند
>
> * اولین مقدار به اولین پارامتر، دومین مقدار به دومین پارامتر و به همین ترتیب اختصاص می‌یابد
> 
> * تغییر ترتیب می‌تواند به نتایج غیرمنتظره‌ای منجر شود

```python
def add(a, b):
    return a + b

# 5 goes to a, 10 goes to b
print(add(5, 10))
```

### Arbitrary arguments (variable-length arguments *args and **kwargs)
### آرگومان‌های دلخواه
> در پایتون، آرگومان‌های دلخواه به ما امکان می‌دهند تعداد متغیری از آرگومان‌ها را به یک تابع ارسال کنیم
>
> این قابلیت زمانی مفید است که از پیش ندانیم به ارسال چه تعداد آرگومان نیاز خواهیم داشت. دو نوع آرگومان دلخواه وجود دارد:
> 
> 1- *args in Python (Non-Keyword Arguments): 
> - آرگومان‌های موقعیتی اضافیِ ارسال‌شده به یک تابع را در یک تاپل جمع‌آوری می‌کند
>
> 2- **kwargs in Python (Keyword Arguments):
> - آرگومان‌های کلیدواژه‌ای اضافیِ ارسال‌شده به یک تابع را در یک دیکشنری جمع‌آوری می‌کند

```python
def multiply(*args):
    result = 1
    for i in args:
        result *= i
    return result

res = multiply(2, 3, 4)
print(res)
```

```python
def myfunc(*args):
    print("Non-Keyword Arguments (*args):")
    print(args)

myfunc('hi', 'peter')
```

```python
# Python program to illustrate
# *args for variable number of arguments
def myFun(*args):
    for arg in args:
        print(arg)


# Driver code with different arguments
myFun('Python', 'is', 'amazing')
```

```python
def area(**kwargs):
    print(kwargs)

area(3, 4)
```

```python
# The above code is wrong because we have to mention the positions
def area(**kwargs):
    print(kwargs)

area(a = 3, b = 4)
```

```python
# Python program to illustrate
# **kwargs for variable number of keyword arguments
def fun(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

# Driver code
fun(course="DSA", platform="GeeksforGeeks", difficulty="easy")
```

```python
def calculate(**kwargs):
    print(f'kwargs are {kwargs}')
    if 'width' in kwargs:
        return kwargs['width'] * kwargs['height']
    if 'radius' in kwargs:
        return kwargs['radius'] * kwargs['radius'] * 3.1415

result = calculate(width = 4, height = 5)
print(f'the area is = {result}')
circle = calculate(radius = 3)
print(f'the circle area is = {circle}') # add :1.2f to circle
```

### The LEGB Rule for Searching Names

از درون یک تابع داخلی هستید x با بازگشت به پرسش بخش قبل، فرض کنید که در حال ارجاع به

می‌گردد x در این حالت، پایتون به ترتیب زیر به دنبال


> **Local:** Python searches for x inside the inner function. If it doesn’t find x there, then it moves to the enclosing scope.
> 
> **Enclosing:** Next, Python searches for x in the enclosing function’s scope. If it’s not found there, then it moves to the global scope.
> 
> **Global:** Python searches for x in the global scope. If it still doesn’t find it, it moves to the built-in scope.
> 
> **Built-in:** Finally, Python searches for x in the built-in scope. If it doesn’t find x there, then it raises a NameError exception. 

```python
x = "global"

def outer():
     def inner():
         print(x)
     inner()


outer()
```

```python
x = "global"

def outer():
     x = "enclosing"
     def inner():
         print(x)
     inner()


outer()
```

```python
x = "global"

def outer():
     x = "enclosing"
     def inner():
         x = "local"
         print(x)
     inner()


outer()
```

### Generators

بسیاری از اشیاء در پایتون از قابلیت پیمایش پشتیبانی می‌کنند؛ برای مثال، پیمایش روی عناصر یک لیست یا خطوط یک فایل

این کار به واسطه‌ی پروتکل تکرارگر انجام می‌شود که روشی کلی برای قابل‌ پیمایش ساختن اشیاء است  

**A generator** 

روشی مناسب — مشابهِ نوشتن یک تابع معمولی — برای ایجاد یک شیء قابل‌تکرار جدید است

برخلاف توابع معمولی که اجرا می‌شوند و در هر بار فراخوانی تنها یک نتیجه برمی‌گردانند

ژنراتورها می‌توانند با متوقف کردن و سپس ازسرگیریِ اجرا در هر بار استفاده، دنباله‌ای از چندین مقدار را بازگردانند

استفاده کنید `yield` از `return` برای ایجاد یک ژنراتور، در بدنه تابع به‌جای کلمه کلیدی


```python
# When you actually call the generator, no code is immediately executed:
def squares(n=10):
    print(f"Generating squares from 1 to {n ** 2}")
    for i in range(1, n + 1):
        yield i ** 2

gen = squares()
print(gen)
```

```python
for x in gen:
    print(x, end=' ')
```

> **Note:**
>
> از آنجا که ژنراتورها خروجی را عنصر به عنصر تولید می‌کنند (در مقابل تولید یک‌جای کل لیست)، این ویژگی می‌تواند به برنامه شما کمک کند تا حافظه کمتری مصرف کند
>
### Why Do We Need Generators?
> * Memory Efficient :
     * مدیریت داده‌های حجیم یا نامحدود بدون بارگذاری کامل آن‌ها در حافظه
>
> * No List Overhead :
     * اقلام را یکی‌یکی ارائه دهید و از ایجاد فهرست کامل خودداری کنید
> 
> * Lazy Evaluation : 
     * مقادیر را تنها در صورت نیاز محاسبه کنید تا عملکرد بهبود یابد
>
> * Support Infinite Sequences : 
     * برای تولید داده‌های نامحدود، مانند دنباله فیبوناچی، ایده‌آل است
>
>
> * Pipeline Processing :
     * ژنراتورها را به هم متصل کنید تا داده‌ها را به‌صورت مرحله‌ای و کارآمد پردازش کنند

### yield vs Return

> **Yield:** 
>
> این دستور در توابع مولد برای ارائه دنباله‌ای از مقادیر در طول زمان به کار می‌رود
>
> عملکرد تابع متوقف شده، مقدار فعلی بازگردانده می‌شود yield با اجرای دستور
>
> این ویژگی به تابع اجازه می‌دهد تا هنگام فراخوانی مجدد، کار خود را از همان نقطه قبلی ادامه دهد؛ امری که آن را به گزینه‌ای ایده‌آل برای تولید کارآمد دنباله‌های بزرگ یا پیچیده تبدیل می‌کند
>
>
> **Return:** 
>
> از آن برای خروج از یک تابع و بازگرداندن یک مقدار نهایی استفاده می‌شود
>
> تابع بلافاصله خاتمه می‌یابد و هیچ وضعیتی حفظ نمی‌شود return به محض اجرای دستور

```python
def fibonacci_numbers(nums):
    x, y = 0, 1
    for _ in range(nums):
        x, y = y, x + y
        yield x

# we can print all in a list
# print(list(fibonacci_numbers(10)))
# or we can create it one by one
fib = fibonacci_numbers(5)
print(next(fib))
# Print next multiple times to see the result!
```
