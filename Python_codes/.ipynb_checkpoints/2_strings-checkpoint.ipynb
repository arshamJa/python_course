> **Strings:** Strings are sequence of characters written inside quotes. It can include letters, numbers, symbols and spaces.

```python
# create a string using double quotes
string1 = "Python programming"
print(string1)
```

```python
# create a string using single quotes
string2 = 'Python programming'
print(string2)
```

> ### Access String Characters in Python

![python_slices](../images/python-slices.png)

> ## Syntax: .[start:stop:step]

```python
string1[0]
```

```python
string1[-1]
```

```python
string1[1:4]
```

```python
string1[:2]
```

```python
string1[-4:-1]
```

```python
string1[1:5:2]
```

### **Note:** Python Strings are Immutable

```python
message = 'Hola Amigos'
message[0] = 'H'
print(message)
```

```python
message = 'Hola Amigos'
new_message = 'h' + message[1:]
print(new_message)
```

### String formatting

```python
name = 'Samael'
greeting = 'welcome to this company.'
print( 'hello {}, {}'.format(name, greeting)  ) # this is used for older versions
```

```python
print(f'hello {name}, {greeting}')
```
