> **Opening a File**
>
> `open()`تابع 
>
>را به عنوان آرگومان می‌پذیرد (mode) برای باز کردن یک فایل استفاده کنیم که مسیر فایل و حالت

```python
f = open("files/open.txt")
print(f)
```

> `read()`تابع
> 
> برای خواندن فایل استفاده می‌کنیم

```python
print(f.read())
```

> **Closing a File**
>
> `file.close()` متد
>
> فایل را می‌بندد و منابع سیستم را آزاد می‌کند
>
>باز شده باشد، بستن آن تضمین می‌کند که تمام تغییرات به‌درستی ذخیره شوند(append)یا الحاق(write) اگر فایل در حالت نوشتن

```python
file = open("files/open.txt", 'r')
content = file.read()
print(content)
file.close()
```

> **Reading Files:** 
>
> پایتون روش‌های متعددی برای خواندن داده‌ها از یک فایل ارائه می‌دهد :
> 
> * `read()`:
     * کل فایل یا تعداد مشخصی از کاراکترها را می‌خواند
>
> * `readline()`:
     * یک خط از فایل را می‌خواند
>
> * `readlines()`:
     *  تمام خطوط فایل را می‌خواند و آن‌ها را در قالب یک لیست برمی‌گرداند

```python
# Reading the entire file
file = open('files/example.txt', 'r')
content = file.read()
print(content)
file.close()

# Reading the first 10 characters
file = open('files/example.txt', 'r')
content = file.read(10)
print(content)
file.close()

# Reading a single line
file = open('files/example.txt', 'r')
line = file.readline()
print(line)
file.close()

# Reading all lines
file = open('files/example.txt', 'r')
lines = file.readlines()
print(lines)
file.close()
```

> **Writing to Files:** writing data to a file in Python can be done using the write() and writelines() methods:
>
> * `write()`:
     * یک رشته را در فایل می‌نویسد
>
> * `writelines()`:
     * فهرستی از رشته‌ها را در فایل می‌نویسد

```python
# Writing a string to a file
file = open('files/example_write.txt', 'w')
file.write('Hello, world!')
file.close()

# Writing a list of strings to a file
lines = ['Hello, world!', 'Python is great!', 'File handling is easy.']
file = open('files/example_write.txt', 'w')
file.writelines(lines)
file.close()
```

> Using **with Statement:**
>
> استفاده کنید with به جای باز و بسته کردن دستی فایل، می‌توانید از دستور
>
> این کار خطر خرابی فایل و نشت منابع را کاهش می‌ده
> 
> حالتی را که فایل در آن باز می‌شود مشخص می‌کند. حالت‌های رایج عبارت‌اند از mode پارامتر :
> * `r` file is opened in **read** mode.
> * `w` file is opened in **write** mode.
> * `a` file is opened in **append** mode.

```python
# Reading the entire file content
with open("files/open.txt", "r") as file:
    content = file.read()
    print(content) # File closes automatically
```

```python
# Reading line by line
with open("files/open.txt", "r") as file:
    for line in file:
        print(line)
```

```python
# Writing a string to a file
with open("files/example.txt", "w") as file:
    file.write("Hello, World!")
```

```python
with open("files/geek.txt", "w") as file:
    file.write("Hello, Python!\n")
    file.write("File handling is easy with Python.")

print("File written successfully")
```

```python
# Writing multiple lines to a file
lines = ["Line 1", "Line 2", "Line 3"]
with open("files/example2.txt", "w") as file:
    for line in lines:
        file.write(line + "\n")
```

```python
# Appending to an existing file
with open("files/example.txt", "a") as file:
    file.write("This line is appended.")
```

### The CSV file
> **Comma Separated Values** 
>
> روشی رایج و ساده برای ذخیره‌سازی داده‌های جدولی است CSV فرمت
>
> باشد .csv آن فایل باید دارای پسوند CSV برای نمایش یک فایل

```python
### Import the CSV module
import csv
```

> Write into CSV files with `csv.writer()`

```python
# newline = creates an empty line between each line we write

with open('files/innovators.csv', 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["SN", "Name", "Contribution"])
    writer.writerow([1, "Linus Torvalds", "Linux Kernel"])
    writer.writerow([2, "Tim Berners-Lee", "World Wide Web"])
    writer.writerow([3, "Guido van Rossum", "Python Programming"])
```

> Writing Multiple Rows with `writerows()`

```python
row_list = [["SN", "Name", "Contribution"],
             [1, "Linus Torvalds", "Linux Kernel"],
             [2, "Tim Berners-Lee", "World Wide Web"],
             [3, "Guido van Rossum", "Python Programming"]]
with open('files/protagonist.csv', 'w') as file:
    writer = csv.writer(file)
    writer.writerows(row_list)
```

### **CSV Files with Custom Delimiters**

> از ویرگول به عنوان جداکننده استفاده می‌شود CSV به‌طور پیش‌فرض، در فایل‌های
>
> ممکن است از جداکننده‌هایی غیر از ویرگول استفاده کنند CSV با این حال، برخی از فایل‌های 
>
> Few popular ones are `|` and `\t`.

```python
data_list = [["SN", "Name", "Contribution"],
             [1, "Linus Torvalds", "Linux Kernel"],
             [2, "Tim Berners-Lee", "World Wide Web"],
             [3, "Guido van Rossum", "Python Programming"]]
with open('files/innovators_pipe_delimiter.csv', 'w', newline='') as file:
    writer = csv.writer(file, delimiter='|')
    writer.writerows(data_list)
```

> Read CSV files with `csv.reader()`

```python
with open('files/innovators.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

> Read CSV files with `csv.DictReader()`

```python
with open("files/people.csv", 'r') as file:
    csv_file = csv.DictReader(file)
    for row in csv_file:
        print(dict(row))
```

### JSON

> **JavaScript Object Notation**
> 
> یک فرمت داده محبوب برای نمایش داده‌های ساختاریافته است
>
> امری رایج است JSON تبادل داده‌ها (ارسال و دریافت) میان سرور و برنامه تحت وب با فرمت

```python
# In Python, JSON exists as a string. For example:
p = '{"name": "Bob", "languages": ["Python", "Java"]}'
```

```python
import json
```

> **Python JSON to dict**
> 
> `json.loads()` متد
>
> تجزیه کنید. این متد یک دیکشنری برمی‌گرداند JSON می‌توانید یک رشته

```python
person = '{"name": "Bob", "languages": ["English", "French"]}'
person_dict = json.loads(person)
print(person_dict)
print('--' * 10)
print(person_dict['languages'])
```

> **Python read JSON file**
>
> `json.load()` متد 
>
> است، استفاده کنید JSON برای خواندن فایلی که حاوی یک شیء 

```python
# Suppose, you have a file named person.json which contains a JSON object.

with open('files/person.json', 'r') as f:
  data = json.load(f)

print(data)
```

> **Python Convert to JSON string**
>
> `json.dumps()` متد 
>
> تبدیل کنید JSON یک دیکشنری را به رشته

```python
person_dict = {
    'name': 'Bob',
    'age': 12,
    'children': None
}
person_json = json.dumps(person_dict)
print(person_json)
```

> **Writing JSON to a file**
>
> `json.dump()` متد 
>
> در یک فایل در پایتون، می‌توانیم استفاده کنیم JSON برای نوشتن

```python
import json
person_dict = {"name": "Bob",
"languages": ["English", "French"],
"married": True,
"age": 32
}

with open('files/person_json.txt', 'w') as json_file:
  json.dump(person_dict, json_file)
```
