### This session we're going to cover:
* If/elif/else condition

> In computer programming, the if statement is a conditional statement. It is used to execute a block of code only when a specific condition is met

```python
# let's see some Boolean operations:
a = 13
b = 33
print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
age = 20
if age >= 18:
    print("Eligible to vote.")
```

```python
age = 10
if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
```

```python
age = 25
if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")
```

![if_elif](../images/if_elif.png)

__[programiz.com/python-programming/if-elif-else](https://www.programiz.com/python-programming/if-elif-else)__

```python
age = 70
is_member = True
if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")
```

### Ternary Conditional Statement

```python
age = 20
s = "Adult" if age >= 18 else "Minor"
print(s)
```

### mini project
*Quiz game* 

```python
print('welcome to quiz game')

playing = input('Do you want to play? ')

if playing != 'yes':
    quit()

print("Ok let's play :) ")
score = 0

answer = input('What does CPU stand for? ')
if answer == 'central processing unit':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

answer = input('What does GPU stand for? ')
if answer == 'graphics processing unit':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

answer = input('What does RAM stand for? ')
if answer == 'random access memory':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

answer = input('What does PSU stand for? ')
if answer == 'power supply':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

print(f'you got {score} questions correct.')
print(f'you got {(score / 4 * 100)}%')
```
