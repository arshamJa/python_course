### This session we're going to cover:

> **Tuple** is a fixed-length, immutable sequence of Python objects which, once assigned,
cannot be changed. The easiest way to create one is with a comma-separated
sequence of values wrapped in parentheses 

```python
tup = (4, 5, 6)
tup
```

> You can convert any sequence or iterator to a tuple by invoking tuple

```python
my_list = [2, 4, 6]
print('list before conversion:\n', my_list)
to_tuple = tuple(my_list)
print('converted list to tuple:\n', to_tuple)
```

```python
tup = tuple('string')
tup
```

```python
nested_tup = (4, 5, 6), (7, 8)
nested_tup
```

```python
nested_tup[0]
```

> While the objects stored in a tuple may be mutable themselves, once the tuple is
created it’s not possible to modify which object is stored in each slot

```python
tup = tuple(['foo', [1, 2], True])
tup[1] = [4, 5]
tup
```

> If an object inside a tuple is mutable, such as a list, you can modify it in place

```python
tup[1].append(3)
```

```python
tup
```

**Tuple methods**

```python
a = (1, 2, 2, 2, 3, 4, 2)
a
```

```python
a.count(2)
```

> **Set** is an unordered collection of unique elements. A set can be created in two ways:
via the set function or via a set literal with curly braces

```python
set([2, 2, 2, 1, 3, 3])
```

> **Sets** support mathematical set operations like union, intersection, difference, and
symmetric difference. Consider these two example sets:

```python
a = {1, 2, 3, 4, 5}
b = {3, 4, 5, 6, 7, 8}
a.union(b) # or we can use a | b
```

```python
a.intersection(b) # or we can use a & b
```

### *Attendance System*

**Story:** Mr. Banner teaches a Python programming class with 10 students. Every morning, he spends 15 minutes checking attendance using a paper register. Sometimes he accidentally marks the wrong student, writes duplicate names, or forgets to count who is absent.

**Why not use Tuple and Set?**

```python
# Permanent class list (Tuple)
class_list = ( "Alice", "Bob", "Charlie", "David", "Emma", "Frank", "Grace", "Henry")

# Students who attended today (Set)
present = {"Alice", "Charlie", "Emma", "Grace", "Henry"}

# Convert tuple to set for calculations
all_students = set(class_list)

# Find absent students
absent = all_students - present

# Display results
print("===== Attendance Report =====")

print("\nClass List:")
print(class_list)

print("\nPresent Students:")
print(present)

print("\nAbsent Students:")
print(absent)

print("\nTotal Students:", len(class_list))
print("Present:", len(present))
print("Absent:", len(absent))

attendance_percentage = (len(present) / len(class_list)) * 100

print("\nAttendance Percentage:", attendance_percentage, "%")
```
