### This session we're going to cover:
* Lists
* Dictionary

> **List** is a built-in data structure used to store an ordered collection of items. They are dynamic, resizable and capable of storing multiple data types.

* *Mutable:* list elements can be changed, updated, added, or removed after the list is created.

* *Ordered:* elements maintain the order in which they are inserted.

* *Index-based:* elements are accessed using their position, starting from index 0.

### Creating List

```python
a_list = [2, 3, 7, None]
a_list
```

```python
a_list[1] = "peekaboo"
a_list
```

```python
gen = range(10)
gen
```

```python
list(gen)
```

### Adding and removing elements

> Elements can be *appended* to the end of the list with the *append* method

```python
a_list.append("dwarf")
```

```python
a_list
```

> Using *insert* you can insert an element at a specific location in the list:

```python
a_list.insert(1, "red")
a_list
```

> The inverse operation to insert is *pop*, which removes and returns an element at a
particular index

```python
a_list.pop(2)
```

```python
a_list
```

> Elements can be removed by value with remove, which locates the first such value and
removes it from the list

```python
a_list.append(3.14)
```

```python
a_list
```

> Elements can be removed by value with remove, which locates the first such value and
removes it from the list

```python
a_list.remove(3.1415)
```

```python
a_list
```

> **NOTE:** difference between pop() vs remove():
> 
* pop takes optional index as value to remove the value, otherwise it removes the last element
* remove takes exactly one argument, and it cannot be empty

### Concatenating and combining lists

> Adding two lists together with + concatenates them

```python
[4, None, "foo"] + [7, 8, (2, 3)]
```

> If you have a list already defined, you can append multiple elements to it using the
extend method

```python
x = [4, None, "foo", 3.14, False]
x.extend([12, 10, 7, (2, 3)])
x
```

### Sorting

> You can sort a list in place (without creating a new object) by calling its sort
function

```python
a = [7, 2, 5, 1, 3]
a.sort()
```

```python
a
```

> sort has a few options that will occasionally come in handy. One is the ability to
pass a secondary sort key—that is, a function that produces a value to use to sort the
objects. For example, we could sort a collection of strings by their lengths

```python
b = ["saw", "small", "He", "foxes", "six"]
b.sort(key=len)
```

```python
b
```

### Slicing

> You can select sections of most sequence types by using slice notation, which in its
basic form consists of start:stop passed to the indexing operator []

```python
seq = [7, 2, 3, 7, 5, 6, 0, 1]
seq[2:5]
```

> Slices can also be assigned with a sequence

```python
seq[1:4] = [12, 13, 14]
seq
```

![list_slice](../images/list_slice.png)

```python
seq[::2]
```

> **Dictionary** stores a collection of key-value pairs, where key and
value are Python objects. One
approach for creating a dictionary is to use curly braces {} and colons to separate
keys and values

```python
d1 = {
    'a': 'some value', 
    'b': [1, 2, 3, 4] 
}
d1
```

> You can access, insert, or set elements using the same syntax as for accessing elements
of a list or tuple

```python
d1[7] = 'an integer'
d1
```

```python
d1[b]
```

```python
# The above code is wrong, instead 
d1['b']
```

> You can delete values using either the del keyword or the pop method (which
simultaneously returns the value and deletes the key)

```python
d1
```

```python
d1['dummy'] = 'some other values'
d1
```

```python
del d1[7]
```

```python
d1
```

```python
ret = d1.pop("dummy")
ret
```

```python
d1
```

> You can merge one dictionary into another using the update method

```python
d1.update({"b": "foo", "c": 12})
```

```python
d1
```
