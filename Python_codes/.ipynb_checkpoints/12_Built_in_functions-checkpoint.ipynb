### Lambda function
> **What is a Lambda Function?**
>
> A lambda function is a small, one-line function defined without a name. It’s called “lambda” because it’s defined using the lambda keyword. These functions can have any number of input parameters, but they have only one expression. The result of this expression is what gets returned.

![lambda](../images/lambda.png)

[Reference: Medium](https://jegananbalagan03.medium.com/lambda-function-in-python-05a0626cc67d)

```python
def short_function(x):
    return x * 2

# Rather than using function we can use Lambda
equiv_anon = lambda x: x * 2
```

```python
x = lambda a, b : a * b
print(x(5, 6))
```

```python
check = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(check(5))   
print(check(-3))  
print(check(0))
```

![map](../images/map.png)

> **map()** function executes a given function to each element of an iterable (such as lists, tuples, etc.)
>
> Syntax: map(function, iterables)
>
> The map() function takes two arguments:
>
> * `function`- a function that is applied to each element of an iterable.
> * `iterables` - iterables such as lists, tuples, etc.

### Benefits of using Map over Loop
* map() is generally faster when working with large collections, as it leverages the C-based implementation of the underlying function.

* loops are typically slower because they involve the interpreter executing the loop statement for each iteration.

```python
numbers = [1,2,3,4]

# returns the square of a number
def square(number):
  return number * number

# apply square() to each item of the numbers list
squared_numbers = map(square, numbers)

# converting to list for printing
result = list(squared_numbers)
print(result)
```

> **map() with Lambda**
>
> In a map() function, we can also use a lambda function instead of a regular function. For example,

```python
numbers = (1, 2, 3, 4)
result = map(lambda x: x*x, numbers)
print(result)

# convert to list and print it
print(list(result))
```

```python
ages = [25, 32, 45, 60, 75]
age_bins = list(map(lambda x: 'young' if x < 35 else ('middle-aged' if x < 60 else 'senior'), ages))
print(age_bins)
```

![zip](../images/zip.jpg)

> **zip** “pairs” up the elements of a number of lists, tuples, or other sequences to create a
list of tuples:

```python
seq1 = ["foo", "bar", "baz"]
seq2 = ["one", "two", "three"]
zipped = zip(seq1, seq2)
list(zipped)
```

![filter](../images/filter.png)

> **filter()** function in Python is used to filter out elements from an iterable (like a list, tuple, etc.) based on a condition. It applies a function to each element of the iterable, and only the elements for which the function returns True are included in the result.
>
> Syntax: filter(function, iterables)
>
> The filter() function takes two arguments:
>
> * `function` - A function that returns either True or False for each element in the iterable..
> * `iterable` - The iterable (like a list or tuple) whose elements you want to filter.

```python
numbers = [1, 2, 3, 4, 5, 6, 7]

def is_even(x):
    return x % 2 == 0
    
even_numbers = filter(is_even, numbers)
print(list(even_numbers))
```

> **filter() with a Lambda**
> 
> You can also use a lambda function with filter() for simpler cases. A lambda function is an anonymous function defined using the lambda keyword.

```python
numbers = [1, 2, 3, 4, 5, 6, 7]

even_numbers = list(filter(lambda x: x % 2 == 0 , numbers))
print(even_numbers)
```

> **reversed** iterates over the elements of a sequence in reverse order:
