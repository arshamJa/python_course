```python
See programiz for more about ascii() and char()

1 Character to ASCII Number (ord)The ord() function takes a single character string and returns its Unicode/ASCII integer value.pythonchar_value = 'g'
ascii_code = ord(char_value)
print(ascii_code)  # Output: 103


2. ASCII Number to Character (chr)The chr() function takes an integer and returns its corresponding character string.pythonascii_code = 103
char_value = chr(ascii_code)
print(char_value)  # Output: g


3. Convert an Entire String to ASCII ValuesTo get the ASCII codes for all characters in a string, you can use a list comprehension:pythontext = "Hello"
ascii_list = [ord(char) for char in text]
print(ascii_list)  # Output: [72, 101, 108, 108, 111]
```
