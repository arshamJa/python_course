### **Object-Oriented Programming (OOP)**
> Is a programming paradigm that organizes code around objects and classes rather than functions and logic alone. It focuses on creating reusable, modular, and scalable code by modeling real-world entities.


![oop](../images/oop.webp)


### Why Use OOP in Python?
> * Promotes **code reusability** through inheritance and composition.
>
> * Increases **modularity** — each class has a specific responsibility.
>
> * Simplifies **maintenance and debugging** by isolating functionality.
>
> * Enables **data encapsulation**, protecting sensitive information.
>
> * Supports **polymorphism**, allowing flexible and dynamic code behavior.

### 1. Class and Object


> A **class** is a blueprint or template that defines the structure and behaviour (attributes and methods) of objects.
>
> An **object** is an instance of a class that holds actual data.
> 
> Classes are used to represent real-world entities like cars, users, or bank accounts in code.
>
> Objects created from these classes carry unique data while sharing the same structure and functionality.

![car](../images/class.webp)

> Consider a Car class: it defines the structure (attributes like color, make, model)
>
> and behavior (methods like start_engine, accelerate) of a car object.

```python
# define a class
class Bank:
    pass

# Creating object from class
account1 = Bank()  # account1 is an object or instance
account2 = Bank()

# Each object has a feature, like name
account1.name = 'Samael'   # name is attribute or property
account2.name = 'Tom'
```

```python
class Parrot:

    # class attribute
    name = ""
    age = 0

# create parrot1 object
parrot1 = Parrot()
parrot1.name = "Blu"
parrot1.age = 10

# create another object parrot2
parrot2 = Parrot()
parrot2.name = "Woo"
parrot2.age = 15

# access attributes
print(f"{parrot1.name} is {parrot1.age} years old")
print(f"{parrot2.name} is {parrot2.age} years old")
```

> `__init__` is a special method (constructor) that initializes attributes when an object is created.
>
> The `self` parameter refers to the instance being created.
> 
> See the below example:

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

> **Instantiating Objects**
>
> Once a class is defined, you can create objects (instances) from it. Let’s create a person1 object from the Person class:

```python
person1 = Person("Alice", 30)
```

> **Accessing Attributes and Methods**
> 
> You can access object attributes and methods using dot notation. For instance, to access the name attribute of person1:

```python
print(person1.name)
```

> To add methods to a class, simply define them within the class block. Let’s add a method to our Person class:

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner       # Instance variable for account holder's name
        self.balance = balance   # Instance variable for account balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print(f"Withdrew {amount}. Remaining balance: {self.balance}")
        else:
            print("Insufficient balance!")

# --- Usage ---
acc1 = BankAccount("Alice", 5000)
acc2 = BankAccount("Bob", 2000)

acc1.deposit(1500)
acc2.withdraw(1000)
```

```python
class Customer:
    def __init__(self, name, account_type):
        """
        Default constructor called automatically when a new object is created.
        Initializes instance variables and can perform setup logic.
        """
        self.name = name
        self.account_type = account_type
        print(f"New customer created: {self.name}, {self.account_type} account")

c1 = Customer("Alice", "Premium")
c2 = Customer("Bob", "Normal")
```

```python
class Area:
    pi = 3.14
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return self.radius * self.radius * Area.pi    # we can type self.pi or Area.pi

c1 = Area(10)
print(c1.area())
```

```python
class Bank:
    bank_name = "National Bank"  # Class variable (shared by all instances)

    def __init__(self, account_holder):
        self.account_holder = account_holder  # Instance variable (unique to each object)
    
    def welcome(self):
        print(f"Welcome {self.account_holder} to {Bank.bank_name}!")

# --- Usage examples ---
# Creating two objects (each with its own instance attribute)
customer1 = Bank("Alice")
customer2 = Bank("Bob")
customer1.welcome()  # Output: Welcome Alice to National Bank!
customer2.welcome()  # Output: Welcome Bob to National Bank!

# Changing the class variable affects all instances
Bank.bank_name = "Global Trust Bank"
customer1.welcome()  # Output: Welcome Alice to Global Trust Bank!
customer2.welcome()  # Output: Welcome Bob to Global Trust Bank!
```
