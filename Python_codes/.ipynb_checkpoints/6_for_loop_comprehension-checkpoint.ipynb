> **for loops** are for iterating over a collection (like a list or tuple) or an iterater. The
standard syntax for a for loop is

```python
languages = ['Swift', 'Python', 'Go']

# start of the loop
for lang in languages:
    print(lang)
    print('-----')
# end of the for loop

print('Last statement')
```

```python
language = 'Python'

# iterate over each character in language
for x in language:
    print(x)
```

```python
my_list = [1, 2, 3, 4]

for i in my_list:
    print(f'{i} * 2 = {i*2}')
```

```python
people = (('tom', 23), ('peter', 32), ('stewie', 9))

for person in people:
    print(person)
```

```python
for name, age in people:
    print(f'{name} is {age} years old.')
```

```python
people = {'peter':(34, 160), 'tom':(40, 170), 'brian':(10, 110)}
for person in people:
    print(person, people[person])
```

> **enumerate()** is used with a for loop to access both the index and the value of each item during iteration.

```python
b = ["eat", "sleep", "repeat"]
for index, value in enumerate(b):
    print (index, value)
```

> **Nested For Loops:** A loop can also contain another loop inside it. These loops are called nested loops.
In a nested loop, the inner loop is executed once for each iteration of the outer loop.

```python
# outer loop 
attributes = ['Electric', 'Fast']
cars = ['Tesla', 'Porsche', 'Mercedes']

for attribute in attributes:
    for car in cars:
        print(attribute, car)
    
    # this statement is outside the inner loop
    print("-----")
```

```python
for i in range(4):
    for j in range(4):
        print((i, j))
```

```python
# Creating math multiplication
for i in range(10):
    for j in range(10):
        print(i, j, end=' ')
    print()
```

```python
rows = int(input("Enter rows:"))
cols = int(input("Enter Cols:"))

for i in range(0, rows):
    for j in range(0, cols):
        print("*", end="")
    print()
```

```python
n = 5
for i in range(n):
    for j in range(i + 1):
        print('*', end=' ')
    print()
```

```python
n = 5
for i in range(n):
    for j in range(i, n):
        print('*', end=' ')
    print()
```

```python
rows = 5
for i in range(1, rows + 1):
    print(' ' * (rows - i) + '*' * (2 * i - 1))
```

```python
n = 5
p = 0

for i in range(n):
    p = 1
    for j in range(i + 1):
        print(p, end=' ')
        p += 1  # p = p + 1
    print()
```

> The **range** function generates a sequence of evenly spaced integers:

```python
range(10)
```

```python
list(range(10))
```

```python
# A start, end, and step (which may be negative) can be given

first_list = list(range(0, 20, 2))

second_list = list(range(5, 0, -1))

print(first_list)
print(second_list)
```

```python
for i in range(0, 10, 2):
    print(i)
```

```python
count_odd = 0
count_even = 0
even_list = []
odd_list = []
for i in range(1, 20):
    if i % 2 == 0:
        even_list.append(i)
        count_even += 1
    else:
        odd_list.append(i)
        count_odd += 1

print(f'even_list = {even_list}, total = {count_even}')
print(f'odd_list = {odd_list}, total = {count_odd}')
```

### **List Comprehensions**
> 
> Syntax: [expr for value in collection if condition]

```python
a = [2, 3, 4, 5]
res = [val ** 2 for val in a]
print(res)
```

```python
a = [1, 2, 3, 4, 5]
res = [val for val in a if val % 2 == 0]
print(res)
```

```python
strings = ["a", "as", "bat", "car", "dove", "python"]
x = []
for string in strings:
    if len(string) > 2:
        x.append(string.upper())

print(x)
```

```python
# Now we use list comprehension
strings = ["a", "as", "bat", "car", "dove", "python"]
result = [x.upper() for x in strings if len(x) > 2]
print(result)
```

### Dictionary comprehension

> **Syntax:** {key: value for (key, value) in iterable if condition}

```python
sq = {x: x**2 for x in range(1, 6)}
print(sq)
```
