### در پایتون if...else دستور 
> به ما امکان می‌دهد تا بر اساس یک شرط مشخص، بلوکی از کد را اجرا کنیم یا از آن صرف‌نظر نماییم `if` در پایتون، دستور
>
> این قابلیت به ما اجازه می‌دهد برنامه‌هایی بسازیم که بسته به شرایط، واکنش‌های متفاوتی نشان می‌دهند

```python
age = 20
if age >= 18:
    print("Eligible to vote.")
```

### (Indentation in Python) تورفتگی در پایتون
> از تورفتگی استفاده می‌کند if پایتون برای تعریف یک بلوک کد، مانند بدنهٔ دستور
> 
> Indentaion is 4 space or 1 tab

```python
age = 10
if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
```

```python
age = 25
if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")
```

```python
age = 70
is_member = True
if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")
```

### Ternary Conditional Statement

```python
age = 20
s = "Adult" if age >= 18 else "Minor"
print(s)
```

### if...else مثال: منطق احراز هویت کاربر با استفاده از

```python
# Username and password stored in database
username_db = "admin"
password_db = "sparrow@123"

# Username and password entered by the user
username = input("Enter username: ")
password = input("Enter password: ") 

# Check if username & password in database matches user's input
if (username == username_db) and (password == password_db):
    print("Welcome back.")
else:
    print("Access denied.")
```

### mini project
*Quiz game*

```python
print('welcome to quiz game')

playing = input('Do you want to play? ')

if playing != 'yes':
    quit()

print("Ok let's play :) ")
score = 0

answer = input('What does CPU stand for? ')
if answer == 'central processing unit':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

answer = input('What does GPU stand for? ')
if answer == 'graphics processing unit':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

answer = input('What does RAM stand for? ')
if answer == 'random access memory':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

answer = input('What does PSU stand for? ')
if answer == 'power supply':
    print('Correct')
    score += 1
else:
    print('Incorrect!')

print(f'you got {score} questions correct.')
print(f'you got {(score / 4 * 100)}%')
```
