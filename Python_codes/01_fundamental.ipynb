### بیایید عملیات ریاضی و متغیر را معرفی کنیم

```python
salary = 5000

rent = 900
food = 650
transport = 250
internet = 180
entertainment = 300
tax = 900

total_expenses = rent + food + transport + internet + entertainment + tax
remaining_money = salary - total_expenses

print(salary)
```

```python
print(total_expenses)
```

```python
print(remaining_money)
```
