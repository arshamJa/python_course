### Mid term exam

> Given Input: colours = ("red", "green", "blue"), replace "green" with "yellow"
>
> Expected Output:
> Original: ('red', 'green', 'blue')
> Modified: ('red', 'yellow', 'blue')

> Given three sets A, B, and C as shown below, write a Python program to find the elements that are in set A but are not present in either set B or set C.
>
> Given Input:
> A = {1, 2, 3, 4, 5, 6, 7, 8}
>
> B = {2, 4, 6}
>
> C = {5, 7, 9}
>
> Expected Output: Result: {1, 3, 8}

> Given a dictionary data, write a Python program to check whether all values in the dictionary are unique. Print True if every value appears only once; otherwise, print False.
>
> Unique Values Check:
>
> Given Input: data = {"a": 1, "b": 2, "c": 3, "d": 2}
>
> Expected Output: False

> Question: create a function that takes a list, uses list comprehension to:
> * place all the 0 values into at the end of the list
> * then order the list from the lowest to the highest value
>
> Given Input: List: [0, 1, 0, 3, 12]
>
> Expected Output: Result: [1, 3, 12, 0, 0]
