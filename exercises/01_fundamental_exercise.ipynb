### Challenge:
> برنامه‌ای به زبان پایتون بنویسید که سن یک فرد را محاسبه کند

الزامات:


* ایجاد کنید و سال تولد خود را به آن اختصاص دهید `birth_year`  متغیری به نام
* ایجاد کنید و سال جاری را به آن اختصاص دهید `current_year`  متغیر دیگری به نام
* سن فرد را با کسر کردن سال تولد از سال جاری محاسبه کنید


### Challenge:
>(BMI) یک برنامه پایتون برای محاسبه شاخص توده بدنی بنویسید


الزامات:

* ایجاد کنید تا وزن یک فرد را بر حسب کیلوگرم ذخیره کند `weight_kg`  متغیری به نام
* ایجاد کنید تا قد فرد را بر حسب متر ذخیره کند `height_m`  متغیری به نام
* را با استفاده از این فرمول محاسبه کنید: (BMI) شاخص توده بدنی
* `BMI = weight_kg / (height_m ** 2)`
